```python
import os
import sys
from pathlib import Path

# Anade la raiz del proyecto al path para importar desde src
ROOT = Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from dotenv import load_dotenv
load_dotenv(ROOT / ".env")

assert os.environ.get("MAPS_KEY"), "Define MAPS_KEY en el archivo .env"
```

```python
from src.config.settings import get_settings
from src.clients.maps_client import GoogleMapsClient
from src.services.places_service import PlacesService

settings = get_settings()
client = GoogleMapsClient(settings)
service = PlacesService(client)
```

```python
# Prueba con un lugar conocido
data = service.search_by_text("Torre del Reformador Guatemala")
data
```

```python
# Extrae los campos mas relevantes del primer candidato
candidates = data.get("candidates", [])
if candidates:
    primero = candidates[0]
    print("Nombre:", primero.get("name"))
    print("Direccion:", primero.get("formatted_address"))
    print("Coordenadas:", primero.get("geometry", {}).get("location"))
    print("Place ID:", primero.get("place_id"))
```
